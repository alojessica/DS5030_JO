We showed that the mean and median could be discovered by minimizing various kinds of loss functions; this is what machine learning is. To make a prediction $\hat{y}(z)$ of $Y$ when $X=z$, minimize the mean squared error:
$$
MSE(\hat{y}(z)) = \dfrac{1}{N} \sum_{i=1}^N \left\lbrace y_i - \hat{y}(z) \right\rbrace^2 \frac{1}{h}k\left(\frac{z-x_i}{h}\right)
$$
Show that the solution to this problem is the LCLS/Naradaya-Watson estimator.

$$
MSE(\hat{y}(z)) = \frac{1}{N}\sum_{i=1}^N \left( y_i - \hat{y}(z) \right)^2 \cdot \frac{1}{h}k\left(\frac{z-x_i}{h}\right)
$$

---
$$
w_i = \frac{1}{h}k\left(\frac{z-x_i}{h}\right), 
\quad a = \hat{y}(z)
$$


$$
MSE(a) = \frac{1}{N}\sum_{i=1}^N w_i (y_i - a)^2
$$

---
$$
\frac{d}{da} MSE(a) 
= \frac{1}{N}\sum_{i=1}^N w_i \cdot 2(a-y_i)
= \frac{2}{N}\sum_{i=1}^N w_i(a-y_i)
$$

---
$$
\sum_{i=1}^N w_i(a-y_i) = 0
$$

---
$$
a\sum_{i=1}^N w_i = \sum_{i=1}^N w_i y_i
\quad \Longrightarrow \quad
a = \frac{\sum_{i=1}^N w_i y_i}{\sum_{i=1}^N w_i}
$$

---
$$
\hat{y}(z) = \frac{\sum_{i=1}^N \dfrac{1}{h}k\!\left(\dfrac{z-x_i}{h}\right) y_i}{\sum_{i=1}^N \dfrac{1}{h}k\!\left(\dfrac{z-x_i}{h}\right)}
$$

---
$$
\hat{y}(z) = \frac{\sum_{i=1}^N K_h(z-x_i) y_i}{\sum_{i=1}^N K_h(z-x_i)},
\quad 
K_h(u) = \frac{1}{h}k\left(\frac{u}{h}\right)
$$

- Write a class or set of functions that implement the LCLS/Naradaya-Watson estimator, using the Silverman plug-in estimate for the conditioning variable $X$ as the bandwidth.
- From one of the course data sets, find two numeric variables of interest, analyze their relationship with the the LCLS/Naradaya-Watson estimator, and discuss your results.

```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from scipy.stats import norm

def silverman_bandwidth(x):
    n = len(x)
    std_x = np.std(x, ddof=1) # Use sample standard deviation
    h = (4 * std_x**5 / (3 * n))**(1/5)
    print(f"Silverman plug-in bandwidth computed: h = {h:.4f}")
    return h

def lcls_nadaraya_watson(x, y, grid_points):
        
    h = silverman_bandwidth(x)
    
    y_hat = np.zeros(len(grid_points))
    x_vals = x.to_numpy()
    y_vals = y.to_numpy()
    
    for i, x_target in enumerate(grid_points):
        u = (x_vals - x_target) / h
        
        weights = norm.pdf(u)
        
        numerator = np.sum(weights * y_vals)
        denominator = np.sum(weights)
        
        if denominator > 0:
            y_hat[i] = numerator / denominator
        else:
            y_hat[i] = np.nan # Assign NaN if no data points are nearby
            
    return y_hat, h
```

```python
if __name__ == '__main__':
    # Load the college completion dataset
    df = pd.read_csv('data/college_completion.csv')

    x_col = 'med_sat_value'
    y_col = 'grad_150_value'

    df_clean = df[[x_col, y_col]].dropna()

    x_variable = df_clean[x_col]
    y_variable = df_clean[y_col]
    
    # grid of points for estimation
    grid = np.linspace(x_variable.min(), x_variable.max(), 100)
    
    # LCLS/Nadaraya-Watson estimator function
    y_hat, h_val = lcls_nadaraya_watson(x=x_variable, y=y_variable, grid_points=grid)
    
    # Plot 
    sns.scatterplot(x=x_variable, y=y_variable, alpha=0.1, label='College Data')
    plt.plot(grid, y_hat, color='orange', linewidth=3, label='N-W Estimate')
    
```
